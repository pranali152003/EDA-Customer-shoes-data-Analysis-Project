```python
# 7. Identify outliers using boxplots or IQR method.
```

```python
for col in ["Price_USD", "Units_Sold", "Revenue_USD"]:
    
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]
    
    print("\nColumn:", col)
    print("Lower limit:", lower)
    print("Upper limit:", upper)
    print("Number of outliers:", len(outliers))
```

```python

fig, ax = plt.subplots(3, 1, figsize=(10, 10))

sns.boxplot(data=df, x="Price_USD", ax=ax[0])
ax[0].set_title("Price Outliers")

sns.boxplot(data=df, x="Units_Sold", ax=ax[1])
ax[1].set_title("Units Sold Outliers")
ax[1].set_xticks(range(1, 21))

sns.boxplot(data=df, x="Revenue_USD", ax=ax[2])
ax[2].set_title("Revenue Outliers")

fig.tight_layout()
plt.show()
```
